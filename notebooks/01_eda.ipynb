```python

import pandas as pd

# 1. Compare the statistics schemas
stats1_cols = pd.read_csv(DATA_DIR / "stats1.csv", nrows=0).columns.tolist()
stats2_cols = pd.read_csv(DATA_DIR / "stats2.csv", nrows=0).columns.tolist()

print("Stats schemas identical:", stats1_cols == stats2_cols)
print("Stats columns:", len(stats1_cols))

if stats1_cols != stats2_cols:
    print("Only in stats1:", sorted(set(stats1_cols) - set(stats2_cols)))
    print("Only in stats2:", sorted(set(stats2_cols) - set(stats1_cols)))

# 2. Inspect IDs and row counts without loading whole files
def inspect_id_file(filename, id_col):
    path = DATA_DIR / filename
    total_rows = 0
    null_ids = 0
    min_id = None
    max_id = None
    unique_ids = set()

    for chunk in pd.read_csv(
        path, usecols=[id_col], chunksize=100_000
    ):
        ids = chunk[id_col]
        total_rows += len(chunk)
        null_ids += int(ids.isna().sum())

        chunk_min = ids.min()
        chunk_max = ids.max()

        if pd.notna(chunk_min):
            min_id = chunk_min if min_id is None else min(min_id, chunk_min)
            max_id = chunk_max if max_id is None else max(max_id, chunk_max)

        unique_ids.update(ids.dropna().unique().tolist())

    print(
        f"{filename}: rows={total_rows:,}, "
        f"unique_{id_col}={len(unique_ids):,}, "
        f"null_ids={null_ids:,}, "
        f"min_id={min_id}, max_id={max_id}"
    )

    return unique_ids

matches_ids = inspect_id_file("matches.csv", "id")
participant_ids = inspect_id_file("participants.csv", "id")
stats1_ids = inspect_id_file("stats1.csv", "id")
stats2_ids = inspect_id_file("stats2.csv", "id")

print("\nID relationship checks")
print("stats1 IDs found in participants:", len(stats1_ids & participant_ids))
print("stats2 IDs found in participants:", len(stats2_ids & participant_ids))
print("stats1/stats2 ID overlap:", len(stats1_ids & stats2_ids))
print("Participant match IDs found in matches:")

participant_match_ids = set()
for chunk in pd.read_csv(
    DATA_DIR / "participants.csv",
    usecols=["matchid"],
    chunksize=100_000
):
    participant_match_ids.update(chunk["matchid"].dropna().unique().tolist())

print(len(participant_match_ids & matches_ids), "/", len(participant_match_ids))
```

```python

import pandas as pd

# A. Inspect match metadata
matches = pd.read_csv(DATA_DIR / "matches.csv")

print("QUEUE DISTRIBUTION")
display(matches["queueid"].value_counts(dropna=False).to_frame("matches"))

print("SEASON DISTRIBUTION")
display(matches["seasonid"].value_counts(dropna=False).sort_index().to_frame("matches"))

print("GAME VERSION DISTRIBUTION — top 15")
display(matches["version"].value_counts(dropna=False).head(15).to_frame("matches"))

print("MATCH DURATION — seconds and minutes")
display(
    matches["duration"].describe().to_frame("seconds")
    .join((matches["duration"] / 60).describe().to_frame("minutes"))
)

# B. Participant counts and player-number distribution
participants = pd.read_csv(
    DATA_DIR / "participants.csv",
    usecols=["id", "matchid", "player", "championid", "role", "position"]
)

participants_per_match = participants.groupby("matchid").size()

print("\nPARTICIPANTS PER MATCH")
display(participants_per_match.value_counts().sort_index().to_frame("number_of_matches"))

print("\nPLAYER NUMBER DISTRIBUTION")
display(participants["player"].value_counts(dropna=False).sort_index().to_frame("rows"))

# C. Load only the target and its participant ID
stats1_target = pd.read_csv(
    DATA_DIR / "stats1.csv", usecols=["id", "win"]
)
stats2_target = pd.read_csv(
    DATA_DIR / "stats2.csv", usecols=["id", "win"]
)
stats_target = pd.concat(
    [stats1_target, stats2_target], ignore_index=True
)

print("\nWIN TARGET DISTRIBUTION")
display(stats_target["win"].value_counts(dropna=False).sort_index().to_frame("rows"))

print("\nWIN PROPORTIONS")
display(
    stats_target["win"].value_counts(normalize=True, dropna=False)
    .sort_index().to_frame("proportion")
)

# D. Verify win labels and how they align with participant player numbers
target_join = participants[["id", "matchid", "player"]].merge(
    stats_target, on="id", how="left", validate="one_to_one", indicator=True
)

print("\nTARGET JOIN STATUS")
display(target_join["_merge"].value_counts().to_frame("rows"))

print("\nWIN RATE BY PLAYER NUMBER")
display(
    target_join.groupby("player", dropna=False)["win"]
    .agg(["count", "mean", "nunique"])
)

print("\nWIN LABELS WITHIN EACH MATCH")
win_counts = target_join.groupby("matchid")["win"].agg(["count", "sum", "nunique"])
display(win_counts["nunique"].value_counts().sort_index().to_frame("number_of_distinct_win_labels"))
```

```python
# 1. Inspect every match that doesn't have exactly 10 participant rows
incomplete_ids = participants_per_match[participants_per_match != 10].index

print(f"Matches with != 10 participants: {len(incomplete_ids)}")
print("\nPARTICIPANT ROWS FOR INCOMPLETE MATCHES")

display(
    target_join[target_join["matchid"].isin(incomplete_ids)]
    .sort_values(["matchid", "player"])
    .merge(
        matches[["id", "queueid", "duration", "seasonid", "version"]]
        .rename(columns={"id": "matchid"}),
        on="matchid",
        how="left",
        validate="many_to_one"
    )
)

# 2. Inspect matches where all available win labels are identical
same_label_ids = win_counts[win_counts["nunique"] == 1].index

print("\nMATCHES WITH ONLY ONE DISTINCT WIN LABEL")
display(
    target_join[target_join["matchid"].isin(same_label_ids)]
    .sort_values(["matchid", "player"])
    .merge(
        matches[["id", "queueid", "duration", "seasonid", "version"]]
        .rename(columns={"id": "matchid"}),
        on="matchid",
        how="left",
        validate="many_to_one"
    )
)

# 3. Find the three participant records without a matching stats row
print("\nPARTICIPANTS WITHOUT MATCHING STATS")
display(
    target_join[target_join["_merge"] == "left_only"]
    .sort_values(["matchid", "player"])
)

# 4. Inspect extremely short matches (under 10 minutes)
print("\nMATCHES UNDER 10 MINUTES")
short_matches = matches[matches["duration"] < 600]

display(
    short_matches[
        ["id", "queueid", "seasonid", "duration", "creation", "version"]
    ].sort_values("duration").head(20)
)

print(f"Total matches under 10 minutes: {len(short_matches)}")
```

```python
# Build one quality summary row per match
match_quality = (
    target_join.groupby("matchid")
    .agg(
        participant_rows=("id", "size"),
        unique_players=("player", "nunique"),
        labeled_rows=("win", "count"),
        distinct_win_labels=("win", "nunique"),
        team1_win_rows=("win", lambda s: (s == 1).sum()),
        team0_win_rows=("win", lambda s: (s == 0).sum()),
    )
    .reset_index()
    .merge(
        matches[["id", "queueid", "seasonid", "duration", "creation", "version"]]
        .rename(columns={"id": "matchid"}),
        on="matchid",
        how="left",
        validate="one_to_one"
    )
)

# Define the quality conditions
match_quality["is_ranked_solo_duo"] = match_quality["queueid"] == 420
match_quality["has_10_participants"] = (
    (match_quality["participant_rows"] == 10)
    & (match_quality["unique_players"] == 10)
)
match_quality["has_10_labels"] = match_quality["labeled_rows"] == 10
match_quality["has_both_outcomes"] = match_quality["distinct_win_labels"] == 2
match_quality["has_five_wins_each"] = (
    (match_quality["team1_win_rows"] == 5)
    & (match_quality["team0_win_rows"] == 5)
)

# Count how many matches pass each condition individually and cumulatively
conditions = [
    ("All matches", pd.Series(True, index=match_quality.index)),
    ("Queue 420", match_quality["is_ranked_solo_duo"]),
    ("Queue 420 + 10 participants", 
     match_quality["is_ranked_solo_duo"] & match_quality["has_10_participants"]),
    ("Plus 10 valid labels",
     match_quality["is_ranked_solo_duo"]
     & match_quality["has_10_participants"]
     & match_quality["has_10_labels"]),
    ("Plus both win outcomes",
     match_quality["is_ranked_solo_duo"]
     & match_quality["has_10_participants"]
     & match_quality["has_10_labels"]
     & match_quality["has_both_outcomes"]),
    ("Plus 5 wins per side",
     match_quality["is_ranked_solo_duo"]
     & match_quality["has_10_participants"]
     & match_quality["has_10_labels"]
     & match_quality["has_both_outcomes"]
     & match_quality["has_five_wins_each"]),
]

summary = pd.DataFrame([
    {"filter": name, "matches_remaining": int(mask.sum())}
    for name, mask in conditions
])

print("CANDIDATE DATASET SIZES")
display(summary)

print("\nWIN-LABEL BALANCE AMONG COMPLETE QUEUE-420 MATCHES")
complete_ranked = match_quality[
    match_quality["is_ranked_solo_duo"]
    & match_quality["has_10_participants"]
    & match_quality["has_10_labels"]
    & match_quality["has_both_outcomes"]
]

display(
    complete_ranked[
        ["team1_win_rows", "team0_win_rows"]
    ].value_counts().rename("matches").to_frame()
)

print("\nDURATION CHECK FOR COMPLETE QUEUE-420 MATCHES")
display(
    complete_ranked["duration"].describe().to_frame("seconds")
    .join((complete_ranked["duration"] / 60).describe().to_frame("minutes"))
)

print(
    "\nComplete ranked matches under 10 minutes:",
    int((complete_ranked["duration"] < 600).sum())
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Keep only participant rows from complete ranked matches
ranked_match_ids = set(complete_ranked["matchid"])

ranked_players = participants[
    participants["matchid"].isin(ranked_match_ids)
].copy()

ranked_players = ranked_players.merge(
    target_join[["id", "win"]],
    on="id",
    how="left",
    validate="one_to_one"
)

# Add champion names
champions = pd.read_csv(DATA_DIR / "champs.csv")

ranked_players = ranked_players.merge(
    champions,
    left_on="championid",
    right_on="id",
    how="left",
    validate="many_to_one"
)

# Count picks and calculate participant-level win rate
champion_summary = (
    ranked_players.groupby(["championid", "name"], dropna=False)
    .agg(
        picks=("win", "size"),
        wins=("win", "sum"),
        win_rate=("win", "mean")
    )
    .reset_index()
)

top_champions = champion_summary.nlargest(15, "picks").sort_values("picks")

fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# 1. Most-picked champions
axes[0].barh(top_champions["name"], top_champions["picks"])
axes[0].set_title("Top 15 Most-Picked Champions")
axes[0].set_xlabel("Participant Picks")
axes[0].set_ylabel("Champion")

# 2. Win rate among those same popular champions
win_rate_sorted = top_champions.sort_values("win_rate")
axes[1].barh(
    win_rate_sorted["name"],
    win_rate_sorted["win_rate"] * 100
)
axes[1].axvline(50, linestyle="--", label="50% reference")
axes[1].set_title("Win Rate of the Top 15 Most-Picked Champions")
axes[1].set_xlabel("Participant Win Rate (%)")
axes[1].set_ylabel("Champion")
axes[1].legend()

plt.tight_layout()
plt.show()

print("TOP 15 CHAMPIONS — SUMMARY")
display(
    top_champions.sort_values("picks", ascending=False)[
        ["name", "picks", "wins", "win_rate"]
    ].assign(win_rate=lambda df: (df["win_rate"] * 100).round(2))
)
```

```python
# Calculate win rate for each participant-based team
player_outcomes = (
    ranked_players[["matchid", "player", "win"]]
    .assign(
        teamid=lambda df: df["player"].map(
            lambda p: 100 if p <= 5 else 200
        )
    )
    .groupby(["matchid", "teamid"], as_index=False)
    .agg(win=("win", "mean"))
)

print("PARTICIPANT-BASED TEAM WIN RATE")
display(
    player_outcomes.groupby("teamid")["win"]
    .agg(["count", "mean", "min", "max"])
)

print("\nTEAMSTATS TEAM IDs")
display(
    ranked_teamstats["teamid"]
    .value_counts(dropna=False)
    .sort_index()
    .to_frame("rows")
)

# Compare teamstats IDs with participant-derived team IDs
team_comparison = ranked_teamstats.merge(
    player_outcomes,
    on=["matchid", "teamid"],
    how="left",
    validate="one_to_one",
    indicator=True
)

print("\nTEAM ID JOIN STATUS")
display(team_comparison["_merge"].value_counts().to_frame("rows"))

print("\nWIN RATE BY TEAMSTATS TEAM ID")
display(
    team_comparison.groupby("teamid")["win"]
    .agg(["count", "mean", "min", "max"])
)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Join team objectives to the provisional team outcome
objective_data = ranked_teamstats.merge(
    player_outcomes,
    on=["matchid", "teamid"],
    how="inner",
    validate="one_to_one"
)

objective_features = {
    "firstblood": "First Blood",
    "firsttower": "First Tower",
    "firstdragon": "First Dragon",
    "firstbaron": "First Baron",
    "firstinhib": "First Inhibitor",
    "firstharry": "First Rift Herald",
}

# Calculate observed win rate for teams that did/did not secure each objective first
objective_rates = []

for col, label in objective_features.items():
    for value in [0, 1]:
        subset = objective_data[objective_data[col] == value]
        objective_rates.append({
            "objective": label,
            "secured_first": "Yes" if value == 1 else "No",
            "teams": len(subset),
            "win_rate_pct": subset["win"].mean() * 100
        })

objective_rates = pd.DataFrame(objective_rates)

fig, ax = plt.subplots(figsize=(12, 6))
objectives = list(objective_features.values())
x = range(len(objectives))
width = 0.36

yes_rates = [
    objective_rates[
        (objective_rates["objective"] == obj)
        & (objective_rates["secured_first"] == "Yes")
    ]["win_rate_pct"].iloc[0]
    for obj in objectives
]
no_rates = [
    objective_rates[
        (objective_rates["objective"] == obj)
        & (objective_rates["secured_first"] == "No")
    ]["win_rate_pct"].iloc[0]
    for obj in objectives
]

ax.bar([i - width/2 for i in x], no_rates, width, label="Did not secure first")
ax.bar([i + width/2 for i in x], yes_rates, width, label="Secured first")
ax.axhline(50, linestyle="--", label="50% reference")
ax.set_xticks(list(x))
ax.set_xticklabels(objectives, rotation=25, ha="right")
ax.set_ylabel("Observed Team Win Rate (%)")
ax.set_title("Observed Win Rate by Team Objective")
ax.legend()
plt.tight_layout()
plt.show()

print("OBJECTIVE WIN-RATE SUMMARY")
display(
    objective_rates.pivot(
        index="objective",
        columns="secured_first",
        values="win_rate_pct"
    ).round(2)
)

print("\nTEAM COUNTS BY OBJECTIVE STATUS")
display(
    objective_rates.pivot(
        index="objective",
        columns="secured_first",
        values="teams"
    )
)
```
